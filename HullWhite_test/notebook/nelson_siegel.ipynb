```python
import numpy as np
import pandas_datareader.data as web
import pandas as pd
```

Get US Treasury yield (daily)

references
- https://fred.stlouisfed.org/series/DGS10/
- https://note.com/scilabcafe/n/ne445b39f1b54

```python
start="2020/4/1"

yield_df = pd.DataFrame([])
for i in [1, 2, 3, 5, 7, 10, 20, 30]:
    print(i)
    ty_tmp = web.DataReader("DGS" + str(i), 'fred',start)#Population, Total for World
    
    yield_df = pd.concat([yield_df,ty_tmp],axis=1).dropna()
```
